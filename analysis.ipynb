```python
#Importing the Dependencies 

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
from sklearn.cluster import KMeans
```

```python
# Data Collection and Analysis
# Loading the data from csv file to a pandas Data Frame 
```

```python
customer_data = pd.read_csv('Mall_Customers.csv')
```

```python
# Displaying the first five rows of the dataset
customer_data.head()
```

```python
#Finding the number of rows and columns in the dataset
customer_data.shape
```

```python
#Getting some information about the dataset
customer_data.info()
```

```python
#Checking for missing values in the dataset
customer_data.isnull().sum()
```

```python
#Choosing the Annual Income Column and Spending Score Column 

X = customer_data.iloc[:, [3, 4]].values
```

```python
print(X)
```

```python
#Choosing the number of clusters
# WCSS -> Within Clusters Sum of Squares            
```

```python
# Finding the WCSS value for different number of clusters

wcss = []

for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
```

```python
# plot an elbow graph

sns.set()
plt.plot(range(1, 11), wcss)
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
# Optimum number of clusters = 5
```

```python
# Training the K-Means Clustering model
kmeans = KMeans(n_clusters=5, init='k-means++', random_state=0)

#return a label for each data point based on the number of clusters
Y = kmeans.fit_predict(X)

print(Y)
```

```python
# Visuallizing all the clusters
```

```python
# plotting all the clusters and their Controls

plt.figure(figsize= (8,8))
plt.scatter(X[Y == 0, 0], X[Y == 0, 1], s = 100, c = 'red', label = 'Cluster 1')
plt.scatter(X[Y == 1, 0], X[Y == 1, 1], s = 100, c = 'blue', label = 'Cluster 2')
plt.scatter(X[Y == 2, 0], X[Y == 2, 1], s = 100, c = 'green', label = 'Cluster 3')
plt.scatter(X[Y == 3, 0], X[Y == 3, 1], s = 100, c = 'cyan', label = 'Cluster 4')
plt.scatter(X[Y == 4, 0], X[Y == 4, 1], s = 100, c = 'magenta', label = 'Cluster 5')

#plot the centroids
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s = 300, c = 'yellow', label = 'Centroids')
plt.title('Customer Groups')
plt.xlabel('Annual Income')
plt.ylabel('Spending Score')
plt.legend()
plt.show()
```
